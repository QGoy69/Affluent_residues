# Runtime

```python
import time
import gc
import statistics
from modules import protein_atoms as PA
from modules import amino_acid_tetras
AAT = amino_acid_tetras.amino_acid_tetras
from modules import tetras_analyser
TA = tetras_analyser.TetraAnalyser

times = []
# Clean memory, reduces measurement variance.
gc.collect()
gc.disable() 

for _ in range(30):
    start = time.process_time()
    analyser = TA([['5aod.pdb','all']], filter = True)
    end = time.process_time()
    gc.enable()
    times.append(end - start)
    
runtime_stat = {
    "mean": statistics.mean(times),
    "median": statistics.median(times),
    "std": statistics.stdev(times),
    "min": min(times),
    "max": max(times),
    "raw": times
}
print(runtime_stat)
```
